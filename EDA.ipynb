```python
numeric_feature = dataset.select_dtypes(exclude=["object"]).columns
catag_feature = dataset.select_dtypes(include=["object"]).columns
```

```python
numeric_feature
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(data=dataset.isnull(), cbar=False, cmap="viridis")
plt.title("Missing value heatmap")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(data=dataset.corr(numeric_only=True), annot=True, cmap="viridis")
plt.title("Correlation heatmap")
plt.show()
```

```python
for col in numeric_feature:
    plt.figure(figsize=(8, 4))
    sns.histplot(dataset[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
for col in numeric_feature:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=dataset[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
for col in catag_feature:
    plt.figure(figsize=(8, 4))
    sns.countplot(x=dataset[col])
    plt.title(f"Countplot of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

```python
sns.pairplot(dataset[numeric_feature])
plt.show()
```

```python
for col in catag_feature:
    print(f"Vlaue count of {col}")
    print(dataset[col].value_counts())
```

```python
for col in numeric_feature:
    plt.figure(figsize=(8, 4))
    sns.violinplot(data=dataset[col])
    plt.show()
```
